```python
import os


def print_tree( root = ".", ignore = None, max_depth = 4, depth = 0 ):
    """
    Print a directory tree from a given root.

    Args:
        root (str): Path to start from.
        ignore (set[str]): Folder names to ignore.
        max_depth (int): Maximum depth to print.
        depth (int): Current recursion depth.
    """
    ignore = ignore or set()

    if depth > max_depth:
        return

    try:
        items = sorted(os.listdir(root))
    except PermissionError:
        return

    for item in items:
        full_path = os.path.join(root, item)

        # Skip ignored directories
        if os.path.isdir(full_path) and item in ignore:
            continue

        indent = "    " * depth
        if os.path.isdir(full_path):
            print(f"{indent}{item}/")
            print_tree(full_path, ignore, max_depth, depth + 1)
        else:
            print(f"{indent}{item}")


print_tree(
        root = r"C:\github\Tree-Canopy-Detection",
        ignore = { ".git",
                   ".codebackup",
                   ".idea",
                   "data1",
                   "data2",
                   "__pycache__",
                   "checkpoints",
                   "checkpoints2",
                   "MANIFEST.in"
                   },
)
```
